```python
def find_letter_folders(data_dir):
    supported_letters = set(config.SUPPORTED_LETTERS)
    letter_folders = []

    for folder_name in sorted(os.listdir(data_dir)):
        folder_path = os.path.join(data_dir, folder_name)

        # Only include directories 
        if not os.path.isdir(folder_path):
            continue

        # Only include supported letter folders
        letter = folder_name.upper()
        if letter in supported_letters:
            letter_folders.append((letter, folder_path))

    print(f"Found {len(letter_folders)} letter folders: {[l for l, _ in letter_folders]}")
    return letter_folders

def find_images_in_folder(folder_path):
    image_extensions = {".jpg"}
    image_paths = []

    for filename in os.listdir(folder_path):
        extension = os.path.splitext(filename)[1].lower()
        if extension in image_extensions:
            image_paths.append(os.path.join(folder_path, filename))

    return sorted(image_paths)
```

## Feature Engineering

```python
def load_already_processed_images(csv_path):
    already_processed = set()

    if not os.path.exists(csv_path):
        return already_processed

    with open(csv_path, "r", newline="") as csv_file:
        reader = csv.DictReader(csv_file)
        for row in reader:
            if "image_path" in row:
                already_processed.add(row["image_path"])

    return already_processed

def process_letter_class(letter, folder_path, sample_limit, already_processed, csv_writer, csv_file):
    image_paths = find_images_in_folder(folder_path)

    stats = {
        "processed": 0,
        "skipped_limit": 0,
        "skipped_no_hand": 0,
        "skipped_error": 0,
        "skipped_already_done": 0
    }

    # Show a progress bar for this letter
    progress_bar = tqdm(
        image_paths,
        desc=f"  Letter {letter}",
        unit="img",
        leave=False
    )

    for image_path in progress_bar:
        # Check if we've hit the sample limit
        if sample_limit > 0 and stats["processed"] >= sample_limit:
            stats["skipped_limit"] += len(image_paths) - stats["processed"] - stats["skipped_already_done"]
            break

        # Skip if already processed (resume mode)
        if image_path in already_processed:
            stats["skipped_already_done"] += 1
            continue

        # Load the image with OpenCV
        try:
            image_bgr = cv2.imread(image_path)
            if image_bgr is None:
                stats["skipped_error"] += 1
                continue

            # Convert BGR to RGB for MediaPipe
            image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
        except Exception as error:
            print(f"\n  Warning: Could not load image {image_path}: {error}")
            stats["skipped_error"] += 1
            continue

        # Run landmark extraction
        try:
            landmarks = extract_landmarks(image_rgb)
        except Exception as error:
            print(f"\n  Warning: Landmark extraction failed for {image_path}: {error}")
            stats["skipped_error"] += 1
            continue

        # Skip images where no hand was detected
        if landmarks is None:
            stats["skipped_no_hand"] += 1
            continue

        # Compute feature vector from landmarks
        try:
            feature_vector = compute_features(landmarks)
        except Exception as error:
            print(f"\n  Warning: Feature computation failed for {image_path}: {error}")
            stats["skipped_error"] += 1
            continue

        # Write this image's features and label to the CSV
        row = {}
        row["image_path"] = image_path
        row["label"] = letter

        feature_names = get_feature_names()
        for feature_name, feature_value in zip(feature_names, feature_vector):
            row[feature_name] = f"{feature_value:.6f}"

        csv_writer.writerow(row)
        stats["processed"] += 1

        # Flush to disk every 100 images so progress is saved if the script crashes
        if stats["processed"] % 100 == 0:
            csv_file.flush()

    return stats
```

```python
def process_images(data_dir, sample=100):
    output = os.path.join(data_dir, "landmarks.csv")
    print(f"Dataset directory: {data_dir}")
    print(f"Output CSV: {output}")

    # Find all letter class folders
    letter_folders = find_letter_folders(data_dir)

    if len(letter_folders) == 0:
        print("ERROR: No letter folders found. Check your dataset path.")
        sys.exit(1)

    # Check for existing CSV (resume capability)
    already_processed = load_already_processed_images(output)
    if len(already_processed) > 0:
        print(f"Resume mode: Found existing CSV with {len(already_processed)} rows. "
            f"Will skip already-processed images.")

    # Prepare CSV headers
    feature_names = get_feature_names()
    fieldnames = ["image_path", "label"] + feature_names

    # Determine if we need to write a header (new file) or append (resume)
    file_mode = "a" if len(already_processed) > 0 else "w"
    write_header = len(already_processed) == 0

    # Open the output CSV and process all letter classes
    with open(output, file_mode, newline="") as csv_file:
        csv_writer = csv.DictWriter(csv_file, fieldnames=fieldnames)

        if write_header:
            csv_writer.writeheader()
            print(f"Created new CSV with {len(fieldnames)} columns "
                f"({len(feature_names)} features + image_path + label)")

        # Track totals across all letters
        total_processed = 0
        total_skipped_no_hand = 0
        total_skipped_error = 0
        per_class_counts = {}

        for letter, folder_path in letter_folders:
            print(f"\nProcessing letter '{letter}' — images in: {folder_path}")

            stats = process_letter_class(
                letter=letter,
                folder_path=folder_path,
                sample_limit=sample,
                already_processed=already_processed,
                csv_writer=csv_writer,
                csv_file=csv_file
            )

            per_class_counts[letter] = stats["processed"]
            total_processed += stats["processed"]
            total_skipped_no_hand += stats["skipped_no_hand"]
            total_skipped_error += stats["skipped_error"]

            print(f"  Letter '{letter}': {stats['processed']} extracted, "
                f"{stats['skipped_no_hand']} skipped (no hand detected), "
                f"{stats['skipped_error']} skipped (errors), "
                f"{stats['skipped_already_done']} skipped (already in CSV)")

    # Print final summary
    print("\n" + "=" * 60)
    print("EXTRACTION COMPLETE")
    print("=" * 60)
    print(f"Total images processed: {total_processed}")
    print(f"Total skipped (no hand): {total_skipped_no_hand}")
    print(f"Total skipped (errors):  {total_skipped_error}")
    print(f"Output saved to: {output}")
    print()
    print("Per-class image counts:")
    for letter, count in sorted(per_class_counts.items()):
        print(f"  {letter}: {count} images")
    print()
    print("Next step: Run scripts/02_train_classifier.py to train the model.")
```
